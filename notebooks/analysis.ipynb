## 2. Clean

Two problems would distort the analysis:

- **Inconsistent author names.** The same author appears with different spacing in their initials (e.g. `J. K. Rowling` vs `J.K. Rowling`), which splits their counts.
- **Repeat titles.** Many books chart in several years. Counting every row would let a few long-running titles dominate genre comparisons.

```python
variants = (raw.assign(normalized=raw['Author'].map(normalize_author))
            .groupby('normalized')['Author'].unique())
variants[variants.map(len) > 1]
```

```python
df = load(DATA)
books = unique_titles(df)
cleaning_report(raw, df), len(books)
```
